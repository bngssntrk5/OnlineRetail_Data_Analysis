```python
import pandas as pd
import matplotlib.pyplot as plt
import os

file_path = os.path.join(os.path.expanduser("~"), "Downloads", "UlkePerformansTablosu.csv")

try:
    df_geo = pd.read_csv(file_path, sep=';')
    
    if df_geo['Total_Revenue'].dtype == 'object':
        df_geo['Total_Revenue'] = df_geo['Total_Revenue'].str.replace(',', '.').astype(float)

    #Ciro Dağılımı (Pasta Grafiği)
    plt.figure(figsize=(10, 6))
    plt.pie(df_geo['Total_Revenue'], labels=df_geo['Country_Group'], autopct='%1.1f%%', 
            startangle=140, colors=['#5dade2', '#f39c12'], explode=(0.1, 0))
    plt.title('Ülke Bazlı Ciro Dağılımı (Global vs UK)', fontsize=16, fontweight='bold', pad=20)
    plt.tight_layout()
    plt.show()

    #Sipariş Hacmi (Bar Grafiği)
    plt.figure(figsize=(10, 6))
    bars = plt.bar(df_geo['Country_Group'], df_geo['Total_Orders'], color='#5dade2', edgecolor='#2e86c1')
    
    for bar in bars:
        yval = bar.get_height()
        plt.text(bar.get_x() + bar.get_width()/2, yval + 500, f'{int(yval):,}', ha='center', va='bottom', fontweight='bold')

    plt.title('Pazar Bazlı Sipariş Hacmi', fontsize=16, fontweight='bold', pad=20)
    plt.xlabel('Ülke Grubu', fontsize=12)
    plt.ylabel('Toplam Sipariş Sayısı', fontsize=12)
    plt.grid(True, linestyle='--', alpha=0.6)
    
    plt.tight_layout()
    plt.show()

except Exception as e:
    print(f"Hata oluştu: {e}")
```
